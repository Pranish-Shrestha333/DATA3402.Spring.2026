1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

we need the initialization method for every class. its to determine or define what data our counter needs to remember. We need the current count and the maximum limit for this task. 

```python
class Counter:
    def __init__(self, max_value):
        self.max_value = max_value
        self.count = 0

    def increment(self):
        # Before adding 1, I must check if we are already at the max_value.
        if self.count < self.max_value:
            self.count += 1
        else:
            print(f"Error: Maximum value of {self.max_value} reached.")

    def reset(self):
        # To reset, I just need to re-assign the count variable back to its starting state (0)
        self.count = 0
        print("Counter has been reset.")
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the parameter of the object. 

```python
import math

class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    def get_area(self):
        raise NotImplementedError()

    def get_perimeter(self):
        raise NotImplementedError()

    def get_perimeter_points(self):
        # Base class template
        raise NotImplementedError("Subclasses must implement get_perimeter_points()")

class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def get_perimeter_points(self):
        # I'll take 4 points per side.
        points = []
        l, w = self.__length, self.__width
        x0, y0 = self._x, self._y
        
        # Points along the top, right, bottom, and left
        for i in range(4):
            points.append((x0 + (i * l/4), y0))          
            points.append((x0 + l, y0 + (i * w/4)))   
            points.append((x0 + l - (i * l/4), y0 + w)) 
            points.append((x0, y0 + w - (i * w/4)))   
        return points[:16]

class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def get_perimeter_points(self):
        # For a circle, I need to use Sine and Cosine to find points on the edge.
        # Gemini showed me that 360 degrees divided by 16 gives the angle step.
        points = []
        for i in range(16):
            angle = (2 * math.pi * i) / 16
            px = self._x + self.__radius * math.cos(angle)
            py = self._y + self.__radius * math.sin(angle)
            points.append((round(px, 2), round(py, 2)))
        return points

class Triangle(Shape):
    def __init__(self, base, height, x, y):
        super().__init__(x, y)
        self.__base = base
        self.__height = height

    def get_perimeter_points(self):
        # We'll treat this as a right-angled triangle starting at (x,y).
        # We divide the 16 points between the 3 sides approx 5 per side.
        points = []
        b, h = self.__base, self.__height
        x, y = self._x, self._y
        
        for i in range(5):
            points.append((x + (i * b/5), y))              # Base
            points.append((x + b - (i * b/5), y + (i * h/5))) # Hypotenuse
            points.append((x, y + h - (i * h/5)))          # Vertical side
        
        return points + [(x,y)] # Adding one more to reach 16

print("Testing Perimeter Points")
rect = Rectangle(10, 10, 0, 0)
circ = Circle(5, 0, 0)

print(f"Rectangle points (first 5): {rect.get_perimeter_points()[:5]}")
print(f"Circle points (first 5): {circ.get_perimeter_points()[:5]}")

# Checking if we actually got 16 points
print(f"\nSuccessfully generated {len(circ.get_perimeter_points())} points for the Circle.")
```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
import math

class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    def is_inside(self, px, py):
        # Base class template
        raise NotImplementedError("Subclasses must implement is_inside()")

class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def is_inside(self, px, py):
        #  I just need to check if the point falls within the boundaries.
        # x must be between self._x and self._x + length.
        in_x = self._x <= px <= (self._x + self.__length)
        in_y = self._y <= py <= (self._y + self.__width)
        return in_x and in_y

class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def is_inside(self, px, py):
        #  If the distance from the point to the center (x, y) is less than the radius, it's inside.
        distance = math.sqrt((px - self._x)**2 + (py - self._y)**2)
        return distance <= self.__radius

class Triangle(Shape):
    def __init__(self, base, height, x, y):
        super().__init__(x, y)
        self.__base = base
        self.__height = height

    def is_inside(self, px, py):
        #This math was hard! Gemini explained the "Area Method."
        # If the point creates 3 smaller triangles that add up to the total area, it's inside.
        def get_tri_area(x1, y1, x2, y2, x3, y3):
            return abs((x1*(y2-y3) + x2*(y3-y1) + x3*(y1-y2)) / 2.0)
        v1 = (self._x, self._y)
        v2 = (self._x + self.__base, self._y)
        v3 = (self._x, self._y + self.__height)

        total_area = get_tri_area(v1[0], v1[1], v2[0], v2[1], v3[0], v3[1])
        area1 = get_tri_area(px, py, v1[0], v1[1], v2[0], v2[1])
        area2 = get_tri_area(px, py, v2[0], v2[1], v3[0], v3[1])
        area3 = get_tri_area(px, py, v1[0], v1[1], v3[0], v3[1])

        # If sum of small areas equals total area, point is inside.
        return math.isclose(total_area, (area1 + area2 + area3))

print("Testing 'is_inside")
rect = Rectangle(10, 10, 0, 0)
circ = Circle(5, 0, 0)

print(f"Is (5,5) inside the 10x10 Rectangle? {rect.is_inside(5, 5)}")  
print(f"Is (15,15) inside the Rectangle? {rect.is_inside(15, 15)}")    
print(f"Is (3,3) inside the Circle (radius 5)? {circ.is_inside(3, 3)}") 
print(f"Is (6,6) inside the Circle? {circ.is_inside(6, 6)}")          
```

The math for checking if a point is inside a triangle was much harder than the rectangle. I had to research the 'Area Method' to solve it. It’s interesting how we can use the sum of smaller triangle areas to determine if a point is within the main boundaries.

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python
import math

class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def v_line(self, x, y, w, **kargs):
        for i in range(x, x + w):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, h, **kargs):
        for i in range(y, y + h):
            self.set_pixel(x, i, **kargs)
            
    def clear_canvas(self):
        # I added this so that when we repaint, 
        # the old version of the drawing is deleted first.
        self.data = [[' '] * self.width for i in range(self.height)]
            
    def display(self):
        print("\n".join(["".join(row) for row in self.data]))

class RasterDrawing:
    def __init__(self, width, height):
        # A drawing needs its own canvas and a way to store multiple shapes.
        # Gemini suggested keeping the width and height so we can clear/reset the canvas.
        self.__width = width
        self.__height = height
        self.__canvas = Canvas(width, height)
        self.__shapes = []

    def add_shape(self, shape):
        # This allows me to build the drawing piece by piece.
        self.__shapes.append(shape)

    def paint(self):
        # First, I clear the canvas so old drawings don't stay behind.
        # Then, I loop through every shape and tell it to paint itself.
        self.__canvas.clear_canvas()
        for s in self.__shapes:
            s.paint(self.__canvas)
        self.__canvas.display()

    def get_shapes(self):
        return self.__shapes

print("Raster Drawing Demo")
drawing = RasterDrawing(20, 40)

rect = Rectangle(10, 15, 2, 5)
drawing.add_shape(rect)

print("First Version of Drawing:")
drawing.paint()

# I wanted to see if I could add a shape later and repaint.
# Gemini showed me that because I store shapes in a list I can just append more.
print("\nModifying Drawing (Adding a Circle)")
circ = Circle(4, 5, 25)
drawing.add_shape(circ)

print("Second Version of Drawing (Modified):")
drawing.paint()
```

```python
import math


# I am redefining these here to make sure they have the __repr__ 
# method that the prof's example showed.

class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length, self.__width = length, width
        
    def paint(self, canvas):
        # Using the h_line and v_line from our previous exercise
        canvas.h_line(self._x, self._y, self.__width)
        canvas.h_line(self._x + self.__length - 1, self._y, self.__width)
        canvas.v_line(self._x, self._y, self.__length)
        canvas.v_line(self._x, self._y + self.__width - 1, self.__length)

    def get_area(self):
        return self.__length * self.__width

    def __repr__(self):
        # I'm following the prof's foo example. 
        # This string allows eval() to recreate the object later.
        return f"Rectangle({self.__length}, {self.__width}, {self._x}, {self._y})"

class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius
        
    def get_area(self):
        return math.pi * (self.__radius ** 2)

    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = (2 * math.pi * i) / 16
            points.append((self._x + self.__radius * math.cos(angle), self._y + self.__radius * math.sin(angle)))
        return points

    def paint(self, canvas):
        for px, py in self.get_perimeter_points():
            try: canvas.set_pixel(int(px), int(py), char='o')
            except: pass

    def __repr__(self):
        # This matches the way I create a Circle in my code
        return f"Circle({self.__radius}, {self._x}, {self._y})"


class RasterDrawing:
    def __init__(self, width, height):
        self.__width, self.__height = width, height
        self.__canvas = Canvas(width, height)
        self.__shapes = []

    def add_shape(self, shape):
        self.__shapes.append(shape)

    def save(self, filename):
        # I'll open a file and write the repr() of every shape.
        # This is exactly like the 'save' method in the foo example.
        with open(filename, "w") as f:
            for s in self.__shapes:
                f.write(repr(s) + "\n")
        print(f"Drawing saved to {filename}")

    def paint(self):
        self.__canvas.clear_canvas()
        for s in self.__shapes:
            s.paint(self.__canvas)
        self.__canvas.display()

def load_raster_drawing(filename, width, height):
    # I used the prof's logic of using eval() 
    # to turn the saved strings back into real Python objects.
    new_drawing = RasterDrawing(width, height)
    with open(filename, "r") as f:
        for line in f:
            if line.strip(): 
                shape_obj = eval(line.strip())
                new_drawing.add_shape(shape_obj)
    return new_drawing

print(" Save/Load Test")
my_art = RasterDrawing(20, 40)
my_art.add_shape(Rectangle(5, 10, 2, 5))
my_art.add_shape(Circle(3, 12, 12))

my_art.save("my_lab_drawing.txt")

print("\nLoading drawing back from file")
reloaded_art = load_raster_drawing("my_lab_drawing.txt", 20, 40)
reloaded_art.paint()
```
